Pyplot functions - these are used to present numbers in pictures .useful for - drawing graphs , data comparison.

plt.plot()-to draw line graph by connecting data points.for visualising relationship between two points.

plt.xlabel('text')- for adding labels on x-axis and y-axis.

plt.title('text')- for giving a head;ine to your chart or graph.

plt.legend ()- it tells about what each line , marker colour is representing in a graph . it only shows if u set a label. and also location can be set like upper left , bottom right and also the font size .

plt.grid()- it adds horizontal and vertical lines in your graph , used for better readability and alignment .

axis limits ( plt.xlim(), plt.ylim())- to control the ranges of x and y axis.

axis ticks( plt.xticks(), plt.yticks())- with this we can gie meaningful labeling to the numeric texts .

plt.show()- for shwoing the final plot .

```python
import matplotlib.pyplot as plt 

#plt.plot(x,y,color='colour_name', linestyle='line_style', linewidth='value', marker='marker symbol', label='label_name') parameters of plot.

months= [1,2,3,4,5]
sales=[100,300,400,500,600]

plt.plot(months,sales, color='pink', linestyle='dashed', linewidth=2, marker ='o', label='sales per month')
plt.xlabel('Months')
plt.ylabel('sales')
plt.title('sales report per month')
plt.legend(loc='upper left',fontsize=10)
plt.grid(color='blue',linestyle='--',linewidth=0.5)
plt.xlim(1,5)
plt.ylim(100,600)
plt.xticks([1,2,3,4,5],['Jan','Feb','Mar','Apr','May'])
plt.show()
```

Visualization tools 
Bar charts-  for categorical comparison of the data.It represents data with rectangular bars , in which the length of the bar is directly proportional to the value it represents.
--> They can be vertical as well as horizontal. for horizontal just use barh.

Pie charts- for porportion illustration.it draws a circle and  show regions with slices.

Histograms- is used to identify how much data is distributed across the ranges(bins).
--> used for continous data not categorical. 
--> divides the data into bins.
-->height of bar= no. of students in that range.

```python
import matplotlib.pyplot as plt
  
product=['A','B','C','D','E']
sales=[100,200,300,400,500]
plt.bar(product, sales , color ='pink', label='sales 2025')
plt.xlabel('product')
plt.ylabel('sales')
plt.title('product sale comparison')
plt.legend(loc='upper left', fontsize='10')
plt.show()
```

```python
import matplotlib.pyplot as plt
#plt.pie(values, labels=label_list, colors=color_list, autopct='%1.1f%)
#autopct = for showing the percentage value on the pie chart.
# %%= for % symbol, 1.1= for showing 1 decimal point, f= for float value.

regions=['North','South','East','West']
revenue=[100,200,300,400]
plt.pie(revenue, labels=regions, colors=['pink','orange','yellow','blue'], autopct='%1.1f%%')
plt.title('Revenue per region')
plt.show()
```

```python
import matplotlib.pyplot as plt
#plt.hist(data, bins= number_of_bins , color='colorname',edgecolor='color')

scores=[13,30,30,44,51,63,45,83,10,14]
plt.hist(scores, bins=10, color='pink', edgecolor='black')
plt.xlabel('score range')
plt.ylabel('number of students')
plt.title('score distribution')
plt.show()
```
