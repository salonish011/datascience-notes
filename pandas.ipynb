```python
import pandas as pd
 
data ={
    'name': ['Alice', 'Bob', 'Charlie'],
    "age":[20,10,30],
    "city":["nagpur","mumbai","delhi"]
    } 
df= pd.DataFrame(data)
print(df)

df.to_csv("output.csv",index=False)
```

Describe Method- It provides a summary of the descriptive statistics for numerical columns in your dataframe.

```python
# step1 - sample data frame
import pandas as pd

data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df=pd.DataFrame(data)
print("sample DataFrame:")
print(df)
print("descriptive statistics:")
print(df.describe())
```

The two outputs are entirely differnet. The describe method have described the data . Count-The no. of non nulls in each column.

Mean- the arithematic average of the value in each column.

Std- (standard deviation)-tells us how much of the values in a column are diff from the average(mean).

For understanding it has two types 

small std - those values which are very close to the average like [49,48,51,52] all are very close to the 50.

large std-like salaries of the employees are very differ and spread out .
Used to measure the consistency like if small std then more consistency.
Used to detect variations like if large std then means large variation in data.

Min - minimum value , smallest value in that column data.

50% also called medain- the middle va;ue of your data when it was sorted. like 22+23/2=22.5 eg - to calculate medium salary paying jobs.
