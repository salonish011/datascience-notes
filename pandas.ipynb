```python
#to filter rows
import pandas as pd

data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df=pd.DataFrame(data)
#to filter rows ona single condition

high_salary=df[df["salary"]>50000]
print(high_salary)

#with multiple conditions using and which returns if all the conditions are true
high_salary=df[(df["salary"]> 50000)& (df["Age"]<50)]
print('Employee salary and age')
print(high_salary)

# using or condition that will return even if one condition is true 
filtered_or= df[(df["Age"]>30) | (df["performance score"] >40)]
print(filtered_or)
```

```python
import pandas as pd
data={
    "Name":["raj","shyam","rishi"],
    "age":[20,24,25],
    "salary":[49000,56000,67000],
}
df=pd.DataFrame(data)
df.sort_values(by="age",ascending=False ,inplace=True)
print("sorting age by descending order:")
print(df)
 #fpr multiple columns we can pass a list of column names to the by parameter of the sort_values() method.
df.sort_values(by=["age","salary"],ascending=[True, False],inplace=True)
print(df)
```

Grouping- dividing dataset  into grps . in the code first the data is grouped and then aggregation is apllied on it.

Grouping in multiple columns- add a list of column u want to aggreagte .
Common aggregation functions- sum(), mean(), max(),min(), std().

```python
import pandas as pd
data={
    "Name":["raj","shyam","rishi","kishan","rahul"],
    "age":[20,24,25,24,20],
    "salary":[49000,56000,67000,78000,89000],
    
}
df= pd.DataFrame(data)
grouped= df.groupby("age")["salary"].sum()
print(grouped)# it will give the sum of salary for each age group

grouped=df.groupby(["age","Name"])["salary"].sum()
print(grouped)
```

Merging and joining - merging - joining  rows of two or more than two dataframes.
syntax - pd.merge(df1,df2, on="column name", how="type of join")

```python
import pandas as pd 
#customer dataframe 
df_customers = pd.DataFrame(
    {'customer_id': [1, 2, 3],
     'Name':['Alice', 'Bob', 'Charlie'],
     }
)

#order dataframe
df_orders=pd.DataFrame({
    'customer_id': [1, 2, 4],
    'orderamount': [200, 240, 367],
    
})
#merge 
df_merged=pd.merge(df_customers,df_orders,on='customer_id',how="inner")
print("inner join")
print(df_merged)

#outer will merge all the rows and will fill missing values with nan
df_merged=pd.merge(df_customers,df_orders,on='customer_id',how="outer")
print("outer join")
print(df_merged)

#left join will keep all the rows from the left dataframe and fill missing values with nan for the right dataframe
df_merged=pd.merge(df_customers,df_orders,on='customer_id',how="left")
print("left join")
print(df_merged)

#right join will keep all the rows from the right dataframe and fill missing values with nan for the left dataframe
df_merged=pd.merge(df_customers,df_orders,on='customer_id',how="right")
print("right join")
print(df_merged)
#cross join will return the cartesian product of both dataframes
```
