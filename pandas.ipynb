Describe Method- It provides a summary of the descriptive statistics for numerical columns in your dataframe.

```python
# step1 - sample data frame
import pandas as pd

data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df=pd.DataFrame(data)
print("sample DataFrame:")
print(df)
print("descriptive statistics:")
print(df.describe())
```

```python
#to filter rows
import pandas as pd

data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df=pd.DataFrame(data)
#to filter rows ona single condition

high_salary=df[df["salary"]>50000]
print(high_salary)

#with multiple conditions using and which returns if all the conditions are true
high_salary=df[(df["salary"]> 50000)& (df["Age"]<50)]
print('Employee salary and age')
print(high_salary)

# using or condition that will return even if one condition is true 
filtered_or= df[(df["Age"]>30) | (df["performance score"] >40)]
print(filtered_or)
```

Modifying data - using two methods

Updating Values using  this .loc[] you can access specific cell , row ,column and can do modification.
syntax- df.loc[row_index,"Column_name"]= new_value
when to use .loc - when u want to update a row, cell or column precisely.

```python
#updating Ram's salary

data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df= pd.DataFrame(data)
df.loc[0,"salary"]=56000
print(df)
```

```python
#if we want to update more than one value of a column 
data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df= pd.DataFrame(data)
#increasing salary by 5%
df["salary"]=df['salary']*1.05
print(df)
```

Removing columns- we use drop method in pandas for it, by this we can delete any row and column from any dataframe.
syntax- df.drop[columns = ["column_name"],inplace= true]
inplace means it will modify the original dataframe directly. 
if  inplace= false, it will return a new dataframe.

```python
data= { "Name":["raj","sahil","rishi","rohit", "ramesh","sachin"],
        "Age":[20,21,22,23,24,25],
        "salary":[49000,50000,56000,67000,47000,76000],
        "performance score":[90,80,45,26,37,27]
}
df= pd.DataFrame(data)
df.drop(columns=["performance score"], inplace=True)
print(df)
df.drop(columns=["Age","salary"],inplace=True)
print(df)
```
